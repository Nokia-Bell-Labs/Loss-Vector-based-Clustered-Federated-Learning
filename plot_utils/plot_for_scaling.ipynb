```python
import os
import matplotlib.pyplot as plt
import pandas as pd
```

```python
directory_path = "../"
results_dir = 'results'
results_sub_dir = 'extensions'

base_dir = os.path.join(directory_path, results_dir)
results_dir_path = os.path.join(base_dir, results_sub_dir)
```

```python
# Get all CSV files from results_dir_path
csv_files = [f for f in os.listdir(results_dir_path) if f.endswith('.csv')]
print(f"Found {len(csv_files)} CSV files:")
for f in csv_files:
    print(f"  - {f}")
```

```python
# Read all CSV files into a dictionary
data_dict = {}
for csv_file in csv_files:
    file_path = os.path.join(results_dir_path, csv_file)
    try:
        df = pd.read_csv(file_path, encoding='utf-8')
    except UnicodeDecodeError:
        df = pd.read_csv(file_path, encoding='latin-1')
    data_dict[csv_file] = df
    print(f"\n{csv_file}:")
    print(f"  Shape: {df.shape}")
    print(f"  Columns: {list(df.columns)}")
```

## Plot: supervised_table_exp203_accuracies.csv

```python
# Plot for supervised_table_exp203_accuracies.csv
df_accuracies = data_dict['supervised_table_exp203_accuracies.csv']
print(df_accuracies.head())
print("\nColumns:", df_accuracies.columns.tolist())

# Extract dataset names and organize columns by dataset
datasets = {}
for col in df_accuracies.columns[1:]:  # Skip first column (algorithm names)
    dataset_name = col.rsplit('_', 1)[0]  # Get dataset name (e.g., 'MNIST' from 'MNIST_50')
    if dataset_name not in datasets:
        datasets[dataset_name] = []
    datasets[dataset_name].append(col)

# Create subplots for each dataset
num_datasets = len(datasets)
fig, axes = plt.subplots(1, num_datasets, figsize=(6*num_datasets, 5))
if num_datasets == 1:
    axes = [axes]

for idx, (dataset_name, cols) in enumerate(sorted(datasets.items())):
    ax = axes[idx]
    
    # Extract client counts and sort columns by client count
    cols_sorted = sorted(cols, key=lambda x: int(x.rsplit('_', 1)[1]))
    client_counts = [int(col.rsplit('_', 1)[1]) for col in cols_sorted]
    
    # Plot each algorithm as a line
    for i, row in df_accuracies.iterrows():
        algorithm = row.iloc[0]  # First column is algorithm name
        values = [row[col] for col in cols_sorted]
        ax.plot(client_counts, values, marker='o', label=algorithm, linewidth=2)
    
    ax.set_xlabel('Number of Clients', fontsize=11)
    ax.set_ylabel('Accuracy', fontsize=11)
    ax.set_title(f'{dataset_name} - Accuracies', fontsize=12, fontweight='bold')
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
